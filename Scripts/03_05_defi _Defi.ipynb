1. Créer un tuple nommé tuple1 qui contient les noms de 2 personnes.

```python
tuple1 = ('Moffo', 'Matho') # ou tuple1 = tuple ({'Moffo', 'Matho'})
tuple1
```

2. Créer un tuple nommé tuple2 qui contient les prénoms d'autant de personne que dans tuple1.

```python
tuple2 = ('Cedric','Derliche')
tuple2
```

3. Former un dictionnaire dict1 qui concatène tuple1 et tuple2

```python
dict1 = {tuple1[0] : tuple2[0], tuple1[1] : tuple2[1]}
dict1
```

4. Créer une copie dict2 de dict1

```python
dict2 = dict1.copy()
dict2
```

5. Ajouter une Personne du nom de Patou et de prénom Alison dans dict2 

```python
dict2['Patou']='Alison'
```

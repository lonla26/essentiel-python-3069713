# 1 Défi

L'idée c'est de construire une fonction, permettant de calculer de manière automatique le salaire d'un employé en tenant compte des heures supplémentaires.
Pour cela, on part de ce postulat :
- On considère les heures supplémentaires, les heures au dessus de 40h.
- Les heures supplémentaires sont rémunérées 1.5 fois le salaire horaire de base

```python
def salaire_employe(salaire_horaire, heures_travaillees):
  if heures_travaillees <= 40:
    salaire = salaire_horaire * heures_travaillees
  else:
    heures_supplementaires = heures_travaillees - 40
    salaire = (salaire_horaire * 40) + (salaire_horaire * 1.5 * heures_supplementaires)
  return salaire
```

```python
salaire_employe(50,30)
```

```python
salaire_employe(50, 45)  # Exemple d'utilisation de la fonction
```

```python
heures_travaillees = 50
tarif_horaire = 100
```

```python
paie = salaire_employe(tarif_horaire, heures_travaillees)
paie
```

```python
print(f"Total à payer : {paie:.2f}", "€")  # Affichage du résultat avec deux décimales
```

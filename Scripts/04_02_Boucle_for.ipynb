## 1 boucle For

**Structure**

for elements in collection :

    instructions

```python
for i in range(4):
  print(i)
```

```python
Liste = ['Un', 'Deux', 'Trois', 'Quatre']
```

```python
for i in Liste:
  print(i)
```

```python
Dict = {'Un': 1, 'Deux': 2, 'Trois': 3, 'Quatre': 4}
```

```python
for key, value in Dict.items():
   print('clé :', key, 'valeur :', value)
```

```python
for item in Liste:
  if item=='Trois':
    break
  print(item)
```

```python
for item in Liste:
  if item=='Deux':
    continue
  print(item)
```

```python
class Vehicule:
  def __init__(self):
    self.est_allume=True
    self.vitesse = 10
  
  def obtenir_vitesse(self):
    return self.vitesse
  
  def deplacement(self):
    if self.est_allume and self.vitesse>0:
      print("Le véhicule se déplace à", self.vitesse, "km/h")
```

```python
v = Vehicule()
```

```python
v.deplacement()
```

```python
class VehiculeMaritime(Vehicule):
  def naviguer(self):
    print("Le véhicule maritime navigue sur l'eau.")
  def deplacement(self):
    if self.est_allume and self.obtenir_vitesse()>0:
      self.naviguer()
```

```python
vm=VehiculeMaritime()
```

```python
vm.deplacement()
```

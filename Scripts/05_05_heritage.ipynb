```python
class Salarie(object):
  def __init__(self, nom, nos):
    self.nom = nom
    self.nos = nos

  def afficher(self):
    print("Nom:", self.nom)
    print("Num salarié:", self.nos)
```

```python
s=Salarie("Alisson", "AP123456")
```

```python
s.afficher()
```

# 1-Défi

1 - Créer un tuple nommé tuple1 qui contient les noms de 2 personnes.

```python
tuple1 = ('Moffo', 'Matho') # ou tuple1 = tuple(['Moffo', 'Matho'])
print(tuple1)
```

2 - Créer un tuple nommé tuple 2 qui contient les prénoms d'autant de personnes que dans le tuple1.

```python
tuple2 = ('Cedric', 'Derliche')
print(tuple2)
```

3 - Former un dictionnaire dict1 qui concatène tuple1 et tuple2

```python
dict1 = {tuple1[i]: tuple2[i] for i in range(len(tuple1))}
print(dict1)
```

4 - Créer une copie dict2 de dict1

```python
dict2 = dict1.copy() #ou dict2 = dict1
print(dict2)
```

5- ajouter une personne du nom Patou et de prenom Alsion dans dict2

```python
dict2['Patou'] = 'Alison'
print(dict2)
```

```python
class CompteBancaire(object):
  def __init__(self, numeroCompte, nom, solde):
    self.numeroCompte = numeroCompte
    self.nom = nom
    self.solde = solde

  def versement(self, montant):
    self.solde += montant
    print(f"Versement de {montant} effectué. Nouveau solde : {self.solde}")

  def retrait(self, montant):
    if montant > self.solde:
      print("Le solde est insuffisant.")
    else:
      self.solde -= montant
      print(f"Retrait de {montant} effectué. Nouveau solde : {self.solde}")

  def agios(self):
    self.solde = self.solde*0.95
    print(f"Des agios ont été appliqués. Nouveau solde : {self.solde}")

  def afficher(self):
    print("Compte numéro :", self.numeroCompte)
    print("Nom du titulaire :", self.nom)
    print("Solde du compte :", self.solde, "€")  
```

```python
monCompte = CompteBancaire("FR123456789", "Alice Dupont", 1000000)
```

```python
monCompte.afficher()
```

```python
monCompte.versement(5000)
```

```python
monCompte.retrait(2000000)
```

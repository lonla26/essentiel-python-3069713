# 1- Paramètres et arguments

```python
def bonjour(prenom):
  print('Bonjour ' + prenom)
```

```python
bonjour('Alisson')
```

```python
bonjour('Damien')
```

```python
def somme(*args):
  x = 0
  for i in args:
    x += i
  print("Ma somme totale est :", x)
```

```python
somme(1,2,3)
```

```python
somme(5)
```

```python
somme(1,2,3,4,5)
```

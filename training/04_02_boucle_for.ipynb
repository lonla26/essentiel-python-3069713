# 1- Boucle For

```python
for i in range(4):
  print(i)
```

```python
Liste = ['Un', 'Deux', 'Trois']
```

```python
for i in Liste:
  print(i)
```

```python
Dict = {'un':1, 'deux':2, 'trois':3}
```

```python
for key, value in Dict.items():
  print('clé :', key, 'valeur :', value)
```

```python
for item in Liste:
  if item=='Deux':
    break
  print(item)
```

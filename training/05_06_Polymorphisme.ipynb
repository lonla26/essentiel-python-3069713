```python
class Vehicule:
  def __init__(self):
    self.est_allume=True
    self.vitesse=10
  def obtenir_vitesse(self):
    return self.vitesse
  def deplacement(self):
    if self.est_allume:
      return "Le véhicule se déplace à la vitesse de {} km/h".format(self.vitesse)
    else:
      return "Le véhicule est éteint."
```

```python
v = Vehicule()
```

```python
v.deplacement()
```

```python
class VehiculeMaritime(Vehicule):
  def naviguer(self):
    print("Le vehicule maritime navique")
  def deplacement(self):
    if self.est_allume and self.obtenir_vitesse() > 0:
      return self.naviguer()
```

```python
vm = VehiculeMaritime()
```

```python
vm.deplacement()
```

# 1-Defi

L'idée est de construire une fonction, permettant de calculer de manière automatique le salaire d'un employé en tenant compte des heureds supplémentaires.
Pour cela, on part de ce postulat:

- On considère les heures supplémentaires, les heures au dessus de 40h.
- Les heures suppléméntaires son rémunérées 1.5 fois le salaire horaire de base.

```python
def salaire_employe(nombre_heures, salaire_horaire):
    if nombre_heures <= 40:
        return nombre_heures * salaire_horaire
    else:
        heures_sup = nombre_heures - 40
        salaire_base = 40 * salaire_horaire
        salaire_sup = heures_sup * salaire_horaire * 1.5
        return salaire_base + salaire_sup
```

```python
salaire_employe(30,50)
```

```python
salaire_employe(45, 50)
```
